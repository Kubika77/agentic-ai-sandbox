## Models Integration With OpenAI, Google Gemini and GROQ

```python
import os
from dotenv import load_dotenv

load_dotenv()

os.environ["OPANAI_API_KEY"]=os.getenv("OPENAI_API_KEY")
os.environ["GOOGLE_API_KEY"]=os.getenv("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
```

### OpenAI model integration

```python
from langchain.chat_models import init_chat_model
model = init_chat_model("gpt-4.1")

model
```

```python
response = model.invoke("Hello how are you?")
response
```

#### Watching the model response content

```python
response.content
```

#### Watchine the tokens usage of the model current response

```python
response.response_metadata["token_usage"]
```

#### Using the langchain openai directly (no use of langchain.chat_models)

```python
from langchain_openai import ChatOpenAI

model = ChatOpenAI(model="gpt-4.1", max_tokens=10)
response = model.invoke("Why Do parrot Parrots talk?")

response
```

```python
response.content
```

```python
response.response_metadata["token_usage"]
```

### Google Gemini model integration

```python
from langchain.chat_models import init_chat_model

model = init_chat_model("google_genai:gemini-2.5-flash-lite")
response = model.invoke("Why do parrots talk?")

response
```

```python
response.content
```

```python
response.usage_metadata
```

#### Using the langchain google genai directly (no use of langchain.chat_models)

```python
from langchain_google_genai import ChatGoogleGenerativeAI

model= ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite")
response = model.invoke("Hello how are you?")

response
```

```python
response.content
```

```python
response.usage_metadata
```

### GROQ model integration

```python
from langchain.chat_models import init_chat_model

model=init_chat_model(model="groq:qwen/qwen3-32b")
response = model.invoke("Hello How are you?")

response
```

```python
response.content
```

```python
response.response_metadata["token_usage"]
```

#### Using the GROQ genai model directly (no use of langchain.chat_models)

```python
from langchain_groq import ChatGroq

model = ChatGroq(model="qwen/qwen3-32b")
response = model.invoke("Hello how are you?", max_tokens=20)

response
```

```python
response.content
```

```python
response.response_metadata["token_usage"]
```
